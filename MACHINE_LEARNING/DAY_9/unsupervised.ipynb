```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

```python
X, y_true = make_blobs(n_samples=500, centers=3, cluster_std=0.60,
random_state=42)
```

```python
df = pd.DataFrame(X, columns=['Feature_1', 'Feature_2'])
```

```python
scaler = StandardScaler()
```

```python
X_scaled = scaler.fit_transform(df)
```

```python
wcss = []
k_range = range(1, 11)
```

```python
for k in k_range:
    Model = KMeans(n_clusters=k, random_state=42)
    Model.fit(X_scaled)
    wcss.append(Model.inertia_)
```

```python
wcss
```

```python
plt.plot(k_range, wcss, marker="x")
```

```python
kmean_final = KMeans(n_clusters=3, random_state=42)
```

```python
cluster_labels = kmean_final.fit_predict(X_scaled)
```

```python
cluster_labels
```

```python
df['cluster'] = cluster_labels
```

```python
sns.scatterplot(x=df['Feature_1'],
                     y=df['Feature_2'],
                     hue=df['cluster'],
                     palette='viridis')
```

```python
from sklearn.datasets import make_moons
```

```python
X, y_true = make_moons(n_samples=500, noise=0.05, random_state=42)
```

```python
from sklearn.cluster import KMeans, DBSCAN
```

```python
df = pd.DataFrame(X, columns=['feature1', 'feature2'])
```

```python
df
```

```python
X_scaled = scaler.fit_transform(df)
```

```python
kmean = KMeans(n_clusters=2, random_state=42)
kmean_labels = kmean.fit_predict(X_scaled)
```

```python
df['kmean_cluster'] = kmean_labels
```

```python
sns.scatterplot(x=df['feature1'], y=df['feature2'], hue=df['kmean_cluster'])
```

```python
dbscam = DBSCAN(eps=0.3, min_samples=5)
dbscam_labels = dbscam.fit_predict(X_scaled)
```

```python
df['cluster_db'] = dbscam_labels
```

```python
sns.scatterplot(x=df['feature1'], y=df['feature2'], hue=df['cluster_db'])
```
