```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
```

```python
df = sns.load_dataset('iris')
```

```python
df.head()
```

```python
df['species'].unique()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X = df.drop('species', axis=1)
y = df['species']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
model_knn = KNeighborsClassifier(n_neighbors=13)
```

```python
model_knn.fit(X_train,y_train)
```

```python
from sklearn.model_selection import cross_val_score
```

```python
model_knn.score(X_test,y_test)
```

```python
from sklearn.svm import SVC
```

```python
model_svm = SVC(gamma='auto')
```

```python
model_svm.fit(X_train,y_train)
```

```python
model_svm.score(X_test,y_test)
```

```python
 #now lets use grid search cv
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
classifer = GridSearchCV((model_svm), {
    "C":[1,10,20,30],
    'kernel':['rbf','linear']
},cv = 5,return_train_score = False)
```

```python
classifer.fit(X,y)
```

```python
classifer.cv_results_
```

```python
results = pd.DataFrame(classifer.cv_results_)
```

```python
results
```

```python
cls = GridSearchCV((model_knn), {
    'n_neighbors': [3,5,7,33,23],
    'weights': ['uniform', 'distance'],
    'algorithm': ['auto', 'ball_tree', 'kd_tree', 'brute']
},cv = 5, return_train_score=False)
```

```python
cls.fit(X,y)
```

```python
rs = cls.cv_results_
```

```python
out = pd.DataFrame(cls.cv_results_)
```

```python
out[['param_algorithm', 'param_n_neighbors',	'param_weights', 'mean_test_score']]
```

```python
from sklearn.model_selection import RandomizedSearchCV
```

```python
rand = RandomizedSearchCV((model_knn), {
    'n_neighbors': [3,5,7,33,23],
    'weights': ['uniform', 'distance'],
    'algorithm': ['auto', 'ball_tree', 'kd_tree', 'brute']
},n_iter=5,cv = 5, return_train_score=False)
```

```python
rand.fit(X,y)
```

```python
result = pd.DataFrame(rand.cv_results_)
```
